**Question 1: Write a short paragraph in your own words explaining what Generative AI (GenAI) is, and give two real-world examples of GenAI applications you have seen or used (such as ChatGPT, DALL-E, or Instagram filters).**

Ans:

Generative AI (GenAI) is a type of artificial intelligence that can create new content such as text, images, music, videos, or code by learning patterns from existing data. Unlike traditional AI, which mainly analyzes or classifies information, GenAI generates original content based on user prompts. Two real-world examples of GenAI that I have used are ChatGPT, which generates human-like text and answers questions, and DALL·E, which creates images from text descriptions. These applications demonstrate how GenAI can assist with creativity, learning, and problem-solving in everyday life.


**Question 2: Create a table comparing Generative AI (GenAI) with Traditional AI by listing at least three key differences in how they work and the types of outputs they produce.**

**Hint: Consider aspects such as data processing, creativity, and output format (fixed outputs vs. newly generated content). **

Ans:

**Traditional AI:**

**Purpose:** Analyzes,classifies or predicts data.

**Data Processing:** Uses input data to make decisions or predictions.

**Creativity:**  Limited to predefined rules or learned patterns.

**Output:** Fixed outputs such as labels, predictions and classification.

**Example Applications:**  Spam detections, Fraud detections and recommendation systems.



**Generative AI:**

**Purpose:** Creates new  content based on learnd patterns.

**Data Processing:** Learns from existing data and generates new text, images, audio,code or videos.

**Creativity:** Produces original and creative content based on user prompts.

**Output:** New and unique outputs such as stories, images,music.

**Example Applications:** ChatGPT, DALL-E, Github Copilot.




**Question 3: List 5 popular apps or features you use daily (like Instagram Reels, Spotify recommendations, Zomato chatbots, YouTube video summaries, or Flipkart product images) and identify which ones likely use GenAI. For each, briefly explain how GenAI might be working behind the scenes.**

Ans:

**Instagram- AI Caption Suggestion:**

One feature on Instagram that likely uses Generative AI is AI-generated caption suggestions. GenAI analyzes the uploaded image, understands its content, and generates personalized captions, hashtags, and emojis. It can also create captions in different tones, such as funny, professional, or motivational, instead of relying on fixed templates.

**Spotify- AI Playlist creation:**

Spotify uses Generative AI to create personalized playlists based on a user's mood, preferences, and listening history. Users can enter prompts such as "Create a relaxing playlist for studying," and the AI generates a playlist that matches the requested mood and activity.

**Zomato- AI Chatbot:**

Zomato's AI chatbot likely uses Generative AI to understand customer queries written in natural language. It can answer questions about orders, recommend restaurants or dishes, resolve delivery issues, and provide personalized responses, making customer support faster and more interactive.

**Youtube- Video Summaries:**

YouTube can use Generative AI to generate short summaries of videos by analyzing their transcripts and identifying the main points. This helps users quickly understand the content before deciding whether to watch the entire video.


**Flipkart- AI Product Descriptions and Images:**

Flipkart likely uses Generative AI to create product descriptions, improve product images, and assist users with natural language search. Instead of using fixed descriptions, the AI generates informative and engaging content that helps customers make better purchasing decisions.
