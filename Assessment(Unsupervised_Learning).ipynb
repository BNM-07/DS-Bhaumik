```python
import pandas as pd
# Load with tab separator to fix the structure
df = pd.read_csv("/content/sample_data/marketing_campaign.csv", sep='\t')
display(df.head())
```

```python
print(df.shape)
print(df.info())
display(df.describe())
```

```python
print(df.isnull().sum())
```

```python
df = df.dropna()
```

```python
# Check duplicates
print(df.duplicated().sum())

# Remove duplicates
df = df.drop_duplicates()
```

```python
print(df.shape)
print(df.isnull().sum())
print(df.duplicated().sum())
```

```python
df.head()
```

```python
## K means Clustering
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Selecting numeric features for clustering
features = ['Income', 'MntWines','MntFruits','MntMeatProducts',
            'MntFishProducts','MntSweetProducts','MntGoldProds']

# Drop rows with missing values in these features specifically
cleaned_df = df.dropna(subset=features)

X = cleaned_df[features]

kmeans = KMeans(n_clusters=4, random_state=42, n_init='auto')
cleaned_df['Cluster'] = kmeans.fit_predict(X)

print("Cluster distribution:")
print(cleaned_df['Cluster'].value_counts())

plt.figure(figsize=(10, 6))
plt.scatter(cleaned_df['Income'], cleaned_df['MntWines'], c=cleaned_df['Cluster'], cmap='viridis', alpha=0.5)
plt.xlabel("Income")
plt.ylabel("Wine Spending")
plt.title("Customer Segments (K-Means)")
plt.colorbar(label='Cluster')
plt.show()
```

```python
import scipy.cluster.hierarchy as sch
import matplotlib.pyplot as plt

features = ['Income', 'MntWines','MntFruits','MntMeatProducts',
            'MntFishProducts','MntSweetProducts','MntGoldProds']

# Fix: Drop rows with missing values in the selected features
X_clustering = df[features].dropna()

plt.figure(figsize=(12, 7))
dendrogram = sch.dendrogram(sch.linkage(X_clustering, method='ward'))
plt.title("Dendrogram (Hierarchical Clustering)")
plt.xlabel("Customers")
plt.ylabel("Euclidean Distance")
plt.show()
```

```python
## PCA

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA

# Use the already loaded dataframe and drop rows with missing values in target features
features = ['Income', 'MntWines','MntFruits','MntMeatProducts',
            'MntFishProducts','MntSweetProducts','MntGoldProds']

X_pca_input = df[features].dropna().drop_duplicates()

# Scaling data for better PCA performance
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X_pca_input)

# K-Means clustering
kmeans = KMeans(n_clusters=4, random_state=42, n_init='auto')
clusters = kmeans.fit_predict(X_scaled)

# PCA transformation to 2 dimensions
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Visualization
plt.figure(figsize=(10, 6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=clusters, cmap='viridis', alpha=0.6)
plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.title("Customer Segments (PCA Visualization)")
plt.colorbar(label='Cluster')
plt.show()
```
