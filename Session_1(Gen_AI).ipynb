**Question 1 : List 3 apps you use daily (like Instagram, Zomato, Spotify) and describe one feature in each that could be powered by a generative AI model instead of traditional rule-based logic.**

Ans:  Here are 3 Apps we use daily and how Generative AI can be used to upgrade these apps:

**1. Instagram - AI Generated Captions:**

In Traditional Rule-Based logic, Instagram may suggest captions from a fixed list or recommend hashtags based on image categories. These suggestions are often generic and do not fully match the photo or the user's style.

But when we use Generative AI, it analyzes the image, understands its context and creates personalized captions, hashtags and emojis. For example when we upload a sunset photo then it generates captions that are of predefined templates. So instead of that predefined templates, Gen AI can adjust the tone to be funny, professional or motivational

Why it's better:  It creates unique, context aware captions instead of fixed    predefined templates


**2. Zomato - Personalized Food Recommendations:**

In traditional rule based logic, Zomato usually recommends restaurants based on previous orders, ratings, location.

But when we integrate Gen AI, it understands natural language requests such as "Suggest a healthy vegetarian dinner under ₹300." It considers your budget, preferences, health goals, and past orders to recommend suitable restaurants and dishes while explaining why they match your request.

Why it's better: It understands user's intent and provides personalized recommendatiions, rather than simple pattern matching.



**3. Spotify - AI Playlist Creation:**

In traditional rule based logic, Spotify recommends songs based on yoir listening history, favorite artists, genres and songs liked by other users.

But when we use Gen Ai, users can create playlists using prompts, for example
"Create a motivational playlist for a workout". It understands the mood, activity, preferences to generate a unique playlist that fits the situation.

Why it's better: It creates a customized playlist based on mood and situation rather than using past listening behavior.




**Question 2: Create a table comparing discriminative models and generative models across at least 4 aspects (definition, main use, example task, example model name).**

Ans:

**Discriminative Models:**
-------------------------

**Definition-**  Learns the relationship between input data and output labels to predict correct class or value.


**Main use-** Classification, regression and prediction tasks.


**Example task-** Predict whether an E-mail is spam or not, classification of images or predicting house prices.


**Example Model-** Logistic Regression, SVM, Random forest, Decison Tree.


**Genarative Models:**
-------------------

**Definition- ** Learns underlying data distribution and generates new data that resembles the training data.

**Main Use**- Creating new content such as images, audio, images and synthetic data.

**Example Task-** Generate realistic images, create music, produce chatbot responses.

**Example Model-** GAN(Genrative Adversarial Network), Diffusion models, VAE(Varible Auto Encoder)






**Question 3: Pick one generative AI application (text generation, image synthesis, music/audio generation, or code generation) and describe a new feature you would add to an app you use, explaining how generative AI would power it.**

Ans:

**Generative AI Application -** Text Generation

**App :** WhatsApp

**New Feature: Smart Conversation Reply Assistant**

I would add a feature called smart conversation reply assistant to whatsApp. This feature would use Generative AI for text generation to help users write personalized and context aware replies.

For example, if someone sends- "Can you join the meeting tomorrow at 10 AM ?"

so instead of basic replies like "Yes", "No" or "Maybe", the AI could generate messages such as - "Yes, I will be there at 10 AM. Looking forward to it." or "Sorry I have another meeting scheduled at that time. Can we reschedule it?.

The AI would understand the conversation, the user's writing style and the context to create natural and personalized responses. It could also adjust the tone to formal, friendly or professional according to the situation.

**How Generative AI works:**

The feature uses Generative AI text model that analyzes previous messages, understands the context and generates new response instead of predefined templates. This makes the conversation faster, more natural and personalized while saving user's time.
