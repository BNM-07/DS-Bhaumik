**Question 1: What do you understand by Natural Language Processing?**

Ans : NLP is a branch of AI that enables computers to understand, interpret and generate human language in a meaningful way. Common applications are Chatbots, Google Translations, Speech recognition(Voice Assitants).

NLP Pipeline:

1.Raw Text

2.Text Preprocessing

3.Stop Word removal

4.Stemming/Lemmitzation

5.Feature Extraction

6.Model Training

7.Evaluation

**Question 2:  What are the steps involved in solving an NLP problem?**

Ans : Steps involved in NLP -:

1.**Problem Definition**: Identify the NLP tasks( Eg: Sentiment Analysis, Text Classification,etc)

2.**Data Collection**:  Gather text data from various sources

3.**Text Preprocessing** : Clean text by converting to lowercase, remove punctuations and special characters and remove unnecessary spaces.

4.**Tokenization**: Split text into words, phrases or sentences.

5.**Remove Stopwords**: Remove common words such as 'a','the', 'is', etc

6.**Stemming/Lemmitization**: Convert the word into its root form.For eg: Running --> Run

7.**Feature Extraction**: Convert text into numerical form using:
--> Bag of Words
--> TF-IDF
--> Word Embeddings(Word2Vec,GloVe)

8.**Model Training**: Train ML or DL model.

9.**Model Evaluation**: Measure Performance.



**Question 3: What is an ensemble method in NLP? With Example.**

Ans: An Ensemble Method is a technique in which multiple machine learning models are combined to improve the accuracy, robustness, and performance of predictions compared to using a single model.

So basically different models may make different errors and combining them can produce better results.

Example:

in sentiment analysis, the predictions of Naive Bayes, SVM, and Logistic Regression can be combined using majority voting to determine whether a review is positive or negative. Common ensemble methods include Bagging, Boosting, and Voting.
